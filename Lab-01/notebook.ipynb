## Experiment-1
Write a Python program to Create a NumPy array using a Python list, Display the array and print the shape and data type of the array.

```python
import numpy as np

# Create array from list
my_list = [10, 20, 30, 40, 50]
arr = np.array(my_list)

print("Array:", arr)
print("Shape:", arr.shape)
print("Data Type:", arr.dtype)
```

## Experiment-2
Write a program to Create a 3x4 matrix of zeros, create a 2x5 matrix of ones and print both matrices and their shapes.

```python
# Matrix of zeros
zeros_mat = np.zeros((3, 4))
print("Zeros Matrix:\n", zeros_mat)
print("Shape of Zeros Matrix:", zeros_mat.shape)

print("\n")

# Matrix of ones
ones_mat = np.ones((2, 5))
print("Ones Matrix:\n", ones_mat)
print("Shape of Ones Matrix:", ones_mat.shape)
```

## Experiment-3
Write a program to Create a NumPy array of numbers from 1 to 12, Reshape the array into a 3x4 matrix and display the original and reshaped array along with their shapes.

```python
# Array from 1 to 12
arr_1_to_12 = np.arange(1, 13)
print("Original Array:", arr_1_to_12)
print("Original Shape:", arr_1_to_12.shape)

print("\n")

# Reshaping
reshaped_arr = arr_1_to_12.reshape(3, 4)
print("Reshaped Matrix:\n", reshaped_arr)
print("Reshaped Shape:", reshaped_arr.shape)
```

## Experiment-4
Write a program to create a NumPy array of 10 random integers, Find the minimum, maximum, mean, and sum of the array.

```python
# Random integers
random_arr = np.random.randint(1, 100, size=10)
print("Random Array:", random_arr)

# Statistical properties
print("Minimum:", random_arr.min())
print("Maximum:", random_arr.max())
print("Mean:", random_arr.mean())
print("Sum:", random_arr.sum())
```

## Experiment-5
Write a Python program to create a Pandas DataFrame using a dictionary containing student names and marks. Display the DataFrame and print the data types (dtypes) of each column.

```python
import pandas as pd

# Dictionary to DataFrame
data = {
    'StudentName': ['Alice', 'Bob', 'Charlie', 'David'],
    'Marks': [85, 92, 78, 88]
}

df_students = pd.DataFrame(data)
print("DataFrame:\n", df_students)
print("\nData Types:\n", df_students.dtypes)
```
